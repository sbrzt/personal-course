## 1 · A function is a rule

A copy shop charges a fee of 1 € for each order, plus 10 cents for each page. The price depends on the number of pages, and only on it: tell the shop how many pages you want, and it can tell you the price.

| pages | 0 | 10 | 50 | 100 |
| --- | --- | --- | --- | --- |
| price (€) | 1 | 2 | 6 | 11 |

A rule like this, which gives exactly one output for each input, is a **function**. Written as a formula:

$$f(x) = 1 + 0.1 \times x$$

| Symbol | Read it as | Means |
| --- | --- | --- |
| $f$ | "f" | the name of the function: here, the shop's price list |
| $x$ | "x", the **input** | the number of pages |
| $f(x)$ | "f of x", the **output** | the price of x pages, in euros |

Read aloud: *"f of x equals one plus zero point one times x."* The brackets in $f(x)$ are not a multiplication: $f(30)$ means "the output of f when the input is 30", here $1 + 0.1 \times 30 = 4$ €. Graphs and books often call the output $y$ and write $y = f(x)$.

Three remarks:

- **Exactly one output for each input.** The shop never charges two prices for the same order. "The articles published in a year" is not a function of the year, because a year has many articles; "the *number* of articles published in a year" is one.
- **A function need not have a formula.** The number of articles *Bibliothecae\.it* published each year (S0.2) is a function of the year, given by a table with no formula behind it.
- **Not every input makes sense.** A number of pages cannot be negative, and the logarithm of S0.2 accepts only positive numbers. The inputs that make sense form the function's **domain**.

You have been using functions all along: `np.log10` is the function "logarithm in base 10", and `len` is the function "number of items". **In code**, `def` makes a new one: its name, its input in brackets, and after `return` its output.

````python
def f(x):
    return 1 + 0.1 * x

print("f(30) =", f(30))
print("for 0, 10, 50 and 100 pages:", f(np.array([0, 10, 50, 100])))
````

Given a numpy array, the same function computes many outputs at once.

## 2 · Reading a graph

The **graph** of a function draws each input with its output as one point: the input along the horizontal axis, the output up the vertical axis. Here is the copy shop's price list:

````python
x = np.linspace(0, 120, 121)
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.plot(x, f(x), color=plots.BLUE)
ax.plot([30, 30, 0], [0, f(30), f(30)], color=plots.ORANGE, ls="--")  # f(30)
ax.plot([0, 90, 90], [10, 10, 0], color=plots.AQUA, ls="--")          # 10 € back
ax.set(xlabel="pages, $x$", ylabel="price in €, $f(x)$",
       xlim=(0, 120), ylim=(0, 14))
plt.show()
````

- **To read an output**, start from the input on the horizontal axis, go up to the graph, then across to the vertical axis. The orange path reads $f(30)$: 30 pages cost 4 €.
- **To read backwards**, start from an output on the vertical axis, go across to the graph, then down. The green path answers "how many pages for 10 €?": 90.
- **Rising and falling.** A graph that climbs from left to right shows a function that **increases**: more input, more output. One that descends shows a function that **decreases**. The steeper the graph, the faster the output changes.

**Reading a graph of data.** Here is the average number of references per article in each year of the journal, with every article counted:

````python
year = bib.articles.set_index("id")["date"].str[:4].astype(int).rename("year")
refs = bibliothecae.refs_per_article(bib).reindex(year.index, fill_value=0)
by_year = refs.groupby(year).mean()          # mean references per article, by year

fig, ax = plt.subplots(figsize=(6.4, 3.2))
ax.plot(by_year.index, by_year.to_numpy(), "o-", color=plots.BLUE)
ax.set(xlabel="year of publication", ylabel="references per article (mean)")
plt.show()
pd.DataFrame({"articles": year.value_counts().sort_index(),
              "mean references": by_year.round(1)})
````

A graph shows the data, not the world; reading it well means asking where the data came from and how many observations stand behind each point. Lesson S1.1 returns to missing data.

## 3 · Straight lines: slope and intercept

The copy shop's graph is a straight line. Every straight line that is not vertical is the graph of a function of this form:

$$y = a + b \times x$$

| Symbol | Read it as | Means |
| --- | --- | --- |
| $a$ | "a", the **intercept** | the value of y when x is 0: where the line crosses the vertical axis |
| $b$ | "b", the **slope** | how much y changes when x grows by 1 |

Read aloud: *"y equals a plus b times x."* For the copy shop, $a = 1$ (the fee) and $b = 0.1$ (the price of each extra page). A positive slope makes the line climb, a negative one makes it fall, and a slope of 0 makes it flat.

**The letters change from book to book; the roles do not.** S0.2 wrote the same kind of line as $x_t = x_0 + a \times t$, where $a$ was the slope. School books often write $y = mx + q$ or $y = mx + b$, and machine learning writes $y = wx + b$, with $b$ the intercept (called the *bias*) and $w$ the slope (the *weight*). To read any of them, look for the roles: the number standing on its own is the intercept, and the number multiplying the input is the slope.

Move the sliders to change the intercept and the slope:

````python
def line(a, b):
    x = np.linspace(-3, 5, 50)
    fig, ax = plt.subplots(figsize=(6.4, 3.6))
    ax.axhline(0, color=plots.MUTED, lw=1)
    ax.axvline(0, color=plots.MUTED, lw=1)
    ax.plot(x, a + b * x, color=plots.BLUE, label=f"y = {a:g} + {b:g} × x")
    ax.plot([1, 2, 2], [a + b, a + b, a + 2 * b], color=plots.ORANGE, ls="--")
    ax.annotate(f"x + 1: y changes by b = {b:g}", (2, a + 1.5 * b),
                xytext=(6, 0), textcoords="offset points", color=plots.ORANGE)
    ax.plot(0, a, "o", color=plots.ORANGE)
    ax.annotate(f"intercept a = {a:g}", (0, a), xytext=(-6, 8),
                textcoords="offset points", ha="right", color=plots.ORANGE)
    ax.set(xlabel="$x$", ylabel="$y$", xlim=(-3, 5), ylim=(-10, 10))
    ax.legend(loc="lower right")
    plt.show()

explore(line, a=(-5.0, 5.0, 0.5, 1.0, "intercept, a"),
        b=(-3.0, 3.0, 0.25, 2.0, "slope, b"))
````

**The slope from two points.** Any two points of a line are enough to find it. The slope is the change in $y$ divided by the change in $x$, often called "rise over run":

$$b = \frac{\Delta y}{\Delta x} = \frac{y_2 - y_1}{x_2 - x_1}$$

| Symbol | Read it as | Means |
| --- | --- | --- |
| $(x_1, y_1)$, $(x_2, y_2)$ | "x one, y one", "x two, y two" | two points of the line |
| $\Delta$ | "delta", a Greek capital D | "the change in" |
| $\Delta y$, $\Delta x$ | "delta y", "delta x" | how much y changes, and how much x changes, between the two points |

Read aloud: *"b equals delta y over delta x, that is, y two minus y one, over x two minus x one."* Once the slope is known, the intercept follows from either point: since $y_1 = a + b \times x_1$, $a = y_1 - b \times x_1$.

**By hand:** for the line through $(2, 5)$ and $(6, 13)$, the slope is $b = (13 - 5) / (6 - 2) = 8 / 4 = 2$, and the intercept is $a = 5 - 2 \times 2 = 1$. So the line is $y = 1 + 2x$. Check it on the other point: $1 + 2 \times 6 = 13$.

````python
x1, y1, x2, y2 = 2, 5, 6, 13
b = (y2 - y1) / (x2 - x1)
a = y1 - b * x1
print(f"slope {b}, intercept {a}; at x = 6 the line gives {a + b * 6}")
````

## 4 · Powers and exponentials

S0.2 met two ways of putting a variable into a power. As functions, they form two families:

$$\text{power:} \quad y = c \times x^k \qquad\qquad \text{exponential:} \quad y = c \times g^x$$

| Symbol | Read it as | Means |
| --- | --- | --- |
| $x$ | "x" | the input |
| $k$ | "k", the exponent | fixed: in a power function, the input is the base |
| $g$ | "g", the growth factor | fixed: in an exponential function, the input is the exponent |
| $c$ | "c" | a fixed number that scales the whole function |

Read aloud: *"y equals c times x to the k"* and *"y equals c times g to the x."* The difference is where the input sits. $x^2$ and $2^x$ look alike, but they are very different functions:

| $x$ | 1 | 2 | 3 | 5 | 10 | 20 |
| --- | --- | --- | --- | --- | --- | --- |
| $x^2$ | 1 | 4 | 9 | 25 | 100 | 400 |
| $2^x$ | 2 | 4 | 8 | 32 | 1,024 | 1,048,576 |

**Power functions in everyday life.**

- The area of a square with side $x$ is $x^2$, and the volume of a cube is $x^3$.
- A square root is a power too: $\sqrt{x} = x^{1/2}$.
- Printing a book has a fixed setup cost, say 1,000 €. Spread over a print run of $x$ copies, it costs $1000 / x$ per copy. That is $1000 \times x^{-1}$, a power function with a negative exponent: the bigger the run, the smaller each copy's share.

**What each family does to its input.** Multiply the input of a power function by some factor, and the output is multiplied by that factor to the power $k$, wherever you start: doubling the side of any square multiplies its area by $2^2 = 4$. Add 1 to the input of an exponential function, and the output is multiplied by $g$: the growth of S0.2.

The exponent sets the shape of a power function:

````python
x = np.linspace(0.05, 3, 200)
fig, ax = plt.subplots(figsize=(6.4, 3.6))
for k, colour_k in [(-1, plots.ORANGE), (0.5, plots.AQUA), (1, plots.MUTED),
                    (2, plots.BLUE)]:
    ax.plot(x, x ** k, color=colour_k, label=f"$x^{{{k:g}}}$")
ax.set(xlabel="$x$", ylabel="$y$", ylim=(0, 6))
ax.legend()
plt.show()
````

With $k$ above 1 the curve bends upwards, ever steeper; between 0 and 1 it bends the other way, ever flatter; with $k = 1$ it is a straight line through 0; and with $k$ below 0 it falls, like the cost per copy.

## 5 · Which axes make which curve straight

Straight lines are easy to read: two numbers, the slope and the intercept, describe them completely. So it is worth knowing the axes on which each family of functions becomes straight.

- **Lines** are straight on ordinary axes.
- **Exponentials** are straight when only the vertical axis is logarithmic. This is called a **semi-log** plot. S0.2 showed why: $\log y = \log c + x \times \log g$ grows by the same amount, $\log g$, at each step.
- **Power functions** are straight when both axes are logarithmic: a **log–log** plot. Take the log of both sides of $y = c \times x^k$, using the rules of logarithms:

$$\log y = \log c + k \times \log x$$

Read aloud: *"log y equals log c plus k times log x."* This is the equation of a straight line, $Y = a + b \times X$, in which $Y = \log y$ and $X = \log x$. Its intercept is $\log c$ and its slope is $k$: **on a log–log plot, the slope of a power function is its exponent.**

| Family | Formula | Straight on | Slope of the straight line |
| --- | --- | --- | --- |
| linear | $y = a + b x$ | ordinary axes | $b$ |
| exponential | $y = c \times g^x$ | a log vertical axis (semi-log) | $\log g$ |
| power | $y = c \times x^k$ | two log axes (log–log) | $k$ |

The same three functions on the three kinds of axes; the straight version of each is in orange:

````python
x = np.linspace(1, 10, 60)
families = {"linear\n$y = 20 + 5x$": 20 + 5 * x,
            "exponential\n$y = 2^x$": 2 ** x,
            "power\n$y = x^3$": x ** 3}
kinds = [("linear", "linear", "ordinary axes"), ("linear", "log", "semi-log"),
         ("log", "log", "log–log")]

fig, grid = plt.subplots(3, 3, figsize=(9, 7))
for i, (name, y) in enumerate(families.items()):
    for j, (xscale, yscale, title) in enumerate(kinds):
        ax = grid[i, j]
        ax.plot(x, y, color=plots.ORANGE if i == j else plots.BLUE, lw=2)
        ax.set_xscale(xscale)
        ax.set_yscale(yscale)
        if i == 0:
            ax.set_title(title)
        if j == 0:
            ax.set_ylabel(name)
fig.tight_layout()
plt.show()
````

Look at the second and third rows on ordinary axes: $2^x$ and $x^3$ both climb from about 1 to about 1,000, and they are hard to tell apart. The log axes separate them at once.

One family overlaps another. A line through 0, $y = b \times x$, is also a power function, with $k = 1$, so it is straight on log–log axes too, with slope 1. A line with an intercept bends on log–log axes, and the larger the intercept compared with $b \times x$, the more it bends.

Move the slider to change the exponent of a power function, and compare the two kinds of axes:

````python
def power_function(k):
    x = np.linspace(0.1, 10, 200)
    fig, axes = plt.subplots(1, 2, figsize=(9, 3.2))
    for ax, scale in zip(axes, ["linear", "log"]):
        ax.plot(x, x ** k, color=plots.BLUE)
        ax.set_xscale(scale)
        ax.set_yscale(scale)
        ax.set(xlabel="$x$", ylabel=f"$x^{{{k:g}}}$",
               title="ordinary axes" if scale == "linear" else "log–log axes")
    fig.tight_layout()
    plt.show()
    print(f"On the log–log axes the line has slope {k:g}: the exponent.")

explore(power_function, k=(-2.0, 3.0, 0.1, 2.0, "exponent, k"))
````

**Reading the exponent off a log–log plot.** Pick two points on the line and count decades, the steps of 10 on each axis. From $x = 10$ to $x = 1{,}000$ is two decades across. If $y$ falls from 500 to 5 meanwhile, two decades down, the slope is $-2 / 2 = -1$, and the function is $c \times x^{-1}$. In general:

$$k = \frac{\log y_2 - \log y_1}{\log x_2 - \log x_1}$$

Read aloud: *"k equals log y two minus log y one, over log x two minus log x one"*: the slope formula of section 3, applied to the logs. Then $c = y_1 / x_1^k$.

Power-law writing has a symbol of its own: $y \propto x^k$, read *"y is proportional to x to the k"*, means $y = c \times x^k$ for some constant $c$ that is left unsaid.

## 6 · Zipf's law in the *Commedia*

In any long text, a few words are everywhere and most words are rare. In the 1930s and 1940s the American linguist George Kingsley Zipf ([Wikipedia](https://en.wikipedia.org/wiki/George_Kingsley_Zipf)) noticed a regularity in this. Rank the words from the most frequent to the least: the second most frequent word appears about half as often as the first, the third about a third as often, and so on. In symbols:

$$f(r) \propto \frac{1}{r^{s}}, \qquad s \approx 1$$

| Symbol | Read it as | Means |
| --- | --- | --- |
| $r$ | "r", the **rank** | 1 for the most frequent word, 2 for the next, and so on |
| $f(r)$ | "f of r" | how many times the word of rank r occurs |
| $\propto$ | "is proportional to" | equal, up to a constant factor |
| $s$ | "s", the exponent | about 1 for many texts |

Read aloud: *"f of r is proportional to one over r to the s, with s about 1."* This is a power function, $f(r) = C \times r^{-s}$, so on log–log axes it should be a straight line with slope $-s$. Here are the words of the *Commedia*, ranked:

````python
words = stats.tokenize(data.commedia())     # every word of the poem, in order
freq = stats.rank_frequency(words)          # counts, most frequent word first
rank = np.arange(1, len(freq) + 1)

fig, ax = plt.subplots(figsize=(6.4, 4))
ax.plot(rank, freq, ".", color=plots.BLUE, ms=3, label="words of the Commedia")
ax.plot(rank, freq[9] * 10 / rank, color=plots.ORANGE, ls="--", label="slope −1")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set(xlabel="rank, $r$", ylabel="occurrences, $f(r)$")
ax.legend()
plt.show()

print(f"{len(words):,} words, {len(freq):,} different ones; "
      f"{np.mean(freq == 1):.0%} of them occur only once")
pd.Series(words).value_counts().head(10).to_frame("occurrences").T
````

From rank 10 to rank 1,000 or so, the points follow a line of slope −1 closely. At both ends they leave it. The most frequent words, such as *e* and *che*, occur less often than the line predicts. At the bottom, thousands of words occur only once or twice. They share the same few counts, so they draw flat steps instead of a line. Zipf's law is a good description of the middle of the ranking, not an exact law: a review of the evidence calls the frequency distribution of words "near-Zipfian" ([Piantadosi, 2014](https://doi.org/10.3758/s13423-014-0585-6)).

> **🔬 Metascience.** Power laws are the oldest regularities of bibliometrics. Alfred Lotka (1926) found that the number of authors who write n papers is about $1/n^2$ times the number who write one ([NIST](https://xlinux.nist.gov/dads/HTML/lotkaslaw.html)): a power function with exponent −2, a straight line of slope −2 on log–log axes. Zipf's law of words has the same shape. Because power laws look straight on log–log axes, a common practice has been to fit a straight line to a log–log plot and report its slope. [Clauset, Shalizi and Newman (2009)](https://doi.org/10.1137/070710111) showed that such least-squares fits can give "substantially inaccurate estimates", and that for some of the 24 real-world datasets they examined, the power law is ruled out altogether. A log–log plot that looks straight is a reason to test for a power law, not a proof of one. Lessons S1.6, S2.7 and S3.6 show how to test it.

## 7 · Which family is it?

Given a curve and no formula, the three kinds of axes tell you which family it belongs to: draw it on all three, and see where it turns straight. This function does exactly that:

````python
def three_axes(x, y, name=""):
    fig, axes = plt.subplots(1, 3, figsize=(9.6, 2.8))
    for ax, (xscale, yscale, title) in zip(axes, kinds):
        ax.plot(x, y, "o-", color=plots.BLUE, ms=3)
        ax.set_xscale(xscale)
        ax.set_yscale(yscale)
        ax.set_title(f"{name}: {title}")
    fig.tight_layout()
    plt.show()

x, curves = check.mystery_curves()     # three curves whose formulas are hidden
for name, y in curves.items():
    three_axes(x, y, name)
````

> **🤖 Why this matters for AI.** A machine-learning model is a function. Its input might be a text, and its output a list of probabilities for the next word. What makes it a *model* is a set of adjustable numbers, its **parameters**, and training means choosing the parameters that make its outputs fit the data. The simplest model of all is the line $y = a + b x$, with two parameters, the intercept and the slope; finding the line that best fits a cloud of points is regression (S5.1). Machine learning writes the same line as $y = wx + b$ (module M3), and a neural network is built from many such pieces: each unit adds up its inputs, each multiplied by its own weight, and passes the sum through a fixed curved function (module M5). Log–log axes are also where the growth of AI is read. Kaplan and colleagues found that the loss of a language model "scales as a power-law with model size, dataset size, and the amount of compute used for training", with some trends spanning more than seven orders of magnitude ([Kaplan et al., 2020](https://arxiv.org/abs/2001.08361v1)). Each of these relations is a straight line on log–log axes, and its slope is the exponent (module M13).
