## 1 · A formula is a sentence

Here is a formula you will meet in lesson S1.4: the mean, or average.

$$\bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i$$

Do not try to understand it yet. Notice only that it reads like a sentence: a subject on the left, the equals sign as its verb, and the rest as what is said about the subject. Read aloud, it says: *"x-bar equals one over n times the sum of x-i, for i from 1 to n."* In plain words: add up all the values, then divide by how many there are. By the end of this lesson you will be able to read every part of it, and much longer formulas too.

Every formula in this course can be read with the same six moves:

1. **Find what is being defined:** usually the symbol on the left of the equals sign.
2. **Name every symbol** on the right.
3. **Find the innermost operation**, the one that is done first.
4. **Read outwards**, one operation at a time.
5. **Try it by hand** on three or four small numbers.
6. **Write it as code**, and check that the code gives what you got by hand.

## 2 · Letters that stand for numbers

A **variable** is a letter that stands for a number, exactly as a variable does in Python. Statistics uses its letters in fairly stable ways, and many of them are Greek. Greek letters are not harder than Latin ones: they are simply more letters. These are the ones you will meet most often.

| Symbol | Read it as | Usually stands for |
| --- | --- | --- |
| x, y | "x", "y" | values in the data |
| n | "n" | how many values there are |
| i, j, k | "i", "j", "k" | a position in a list (an *index*) |
| μ | "mu" | the mean of a whole population |
| σ | "sigma" | a standard deviation (how spread out values are) |
| p | "p" | a probability or a proportion |
| α, β, θ, λ | "alpha", "beta", "theta", "lambda" | the settings of a model, which a method estimates from data (*parameters*) |
| ρ, κ | "rho", "kappa" | a correlation, an agreement score |

Two marks change a letter's meaning. A **bar** over it, as in $\bar{x}$ ("x-bar"), means the mean of the values. A **hat**, as in $\hat{p}$ ("p-hat"), means "estimated from data": $p$ is the true proportion, which we usually cannot see, and $\hat{p}$ is our estimate of it. Part 3 is all about the gap between the two.

## 4 · The sum sign Σ: "add them all up"

The capital Greek letter Σ (sigma, the Greek S, as in *sum*) means "add up". It has four parts:

$$\sum_{i=1}^{n} x_i$$

| Part | Read it as | What it says |
| --- | --- | --- |
| Σ | "the sum..." | add up the terms that follow |
| i = 1, below | "...for i from 1..." | start with i = 1 |
| n, above | "...to n..." | stop after i = n |
| $x_i$, after | "...of x-i" | the term added each time, with i set to the current position |

Read aloud: *"the sum of x-i, for i from 1 to n."* It is a short way of writing $x_1 + x_2 + \dots + x_n$.

**By hand**, for our six essays: 170 + 53 + 10 + 102 + 22 + 13 = 370.

**In code**, a sum is a loop that keeps a running total. Numpy also has it built in:

````python
total = 0
for i in range(1, n + 1):          # i = 1, 2, ..., n, as in the formula
    total = total + x[i - 1]       # x_i is x[i - 1] in Python
print("loop:", total, "   numpy:", x.sum())
````

Move the slider to build the sum one term at a time:

````python
def running_sum(k):
    colours = [plots.BLUE if i < k else plots.MUTED for i in range(n)]
    fig, ax = plt.subplots(figsize=(6.4, 3.2))
    ax.bar(range(1, n + 1), x, color=colours)
    ax.set(xlabel=r"$i$ (position in the list)", ylabel=r"$x_i$ (references)",
           title=f"Sum of the first {k} terms = {x[:k].sum()}")
    plt.show()

explore(running_sum, k=(1, n, 1, 3, "terms added, k"))
````

Where the brackets and powers go changes the result. Read these four carefully:

$$\sum_{i=1}^{n} x_i^2 \qquad \Big(\sum_{i=1}^{n} x_i\Big)^2 \qquad \sum_{i=1}^{n} (x_i - 50) \qquad \sum_{i=1}^{n} 1 = n$$

1. "The sum of x-i squared": square each value first, then add the squares.
2. "The square of the sum": add first, then square the total.
3. "The sum of x-i minus 50": take 50 from each value, then add the results.
4. "The sum of 1": adding 1 once for every value counts the values, so it equals n.

By hand, for the values 1, 2 and 3: the sum of squares is 1 + 4 + 9 = 14, but the square of the sum is 6² = 36.

## 5 · The product sign Π: "multiply them all"

The capital Greek letter Π (pi, the Greek P, as in *product*) works like Σ, but multiplies instead of adding:

$$\prod_{i=1}^{n} x_i = x_1 \times x_2 \times \dots \times x_n$$

Read aloud: *"the product of x-i, for i from 1 to n."* By hand, for 0.5, 0.2 and 0.9: 0.5 × 0.2 × 0.9 = 0.09.

Products matter more than they seem. In Part 2, the chance that several independent things all happen is the product of their chances. In Part 3, the *likelihood* of a whole dataset is a product over its values. And products of many numbers below 1 shrink very fast:

````python
p = np.full(400, 0.1)        # 400 chances of 1 in 10
print("product of the first 10: ", np.prod(p[:10]))
print("product of the first 100:", np.prod(p[:100]))
print("product of all 400:      ", np.prod(p))
````

The last answer is not really 0: it is $10^{-400}$, a number too small for the computer to store, so it is rounded to 0. This is called *underflow*. Lesson S0.2 shows how logarithms avoid it, by turning products into sums.

## 6 · Long formulas: read them from the inside out

A few more symbols, all familiar from school:

| Symbol | Read it as | Means | In Python |
| --- | --- | --- | --- |
| $x^2$ | "x squared" | x × x | `x ** 2` |
| $\sqrt{x}$ | "the square root of x" | the number that, squared, gives x | `np.sqrt(x)` |
| $\lvert x \rvert$ | "the absolute value of x" | x without its sign: \|−3\| = 3 | `abs(x)` |
| $\frac{a}{b}$ | "a over b" | a divided by b; the bar also groups everything above it and everything below it | `a / b` |

The order of operations is the one you know, and the same as in Python: brackets first, then powers and roots, then multiplication and division, then addition and subtraction.

Now a long formula. This is the **standard deviation**, which lesson S1.5 explains. Here we only read it and compute it:

$$s = \sqrt{\frac{\sum_{i=1}^{n} \left(x_i - \bar{x}\right)^2}{n - 1}}$$

Read it from the inside out:

1. $\bar{x}$ is the mean: add up the values and divide by n.
2. $x_i - \bar{x}$: take the mean away from each value. These are the *deviations*.
3. $(x_i - \bar{x})^2$: square each deviation.
4. $\sum$: add the squares up.
5. $\dots / (n - 1)$: divide by one less than the number of values.
6. $\sqrt{\dots}$: take the square root of the result.

Read aloud: *"s equals the square root of the sum of the squared differences between each x-i and x-bar, divided by n minus one."*

Move the slider to change the first value, $x_1$, and watch every step change with it:

````python
def sd_steps(x1):
    v = x.astype(float).copy()
    v[0] = x1
    m = v.mean()
    steps = pd.DataFrame({"x_i": v, "x_i − x̄": v - m, "(x_i − x̄)²": (v - m) ** 2})
    steps.index = pd.Index(range(1, n + 1), name="i")  # i = 1, ..., n
    display(steps.round(1))
    total = ((v - m) ** 2).sum()
    s_now = np.sqrt(total / (n - 1))
    print(f"x̄ = {m:.1f}   sum of squares = {total:,.1f}   "
          f"divided by n − 1 = {total / (n - 1):,.1f}   s = {s_now:.1f}")

explore(sd_steps, x1=(0, 400, 10, 170, "x₁"))
````

## 7 · Conditions and counting

Formulas often contain comparisons:

| Symbol | Read it as | In Python |
| --- | --- | --- |
| $<$, $>$ | "less than", "greater than" | `<`, `>` |
| $\le$, $\ge$ | "at most", "at least" | `<=`, `>=` |
| $\ne$ | "not equal to" | `!=` |
| $\approx$ | "approximately equal to" | `np.isclose(a, b)` |

A neat trick turns a condition into a number. The **indicator** $\mathbb{1}[\,\text{condition}\,]$ ("one if") equals 1 when the condition is true and 0 when it is false. Adding indicators counts how often the condition holds:

$$\sum_{i=1}^{n} \mathbb{1}[x_i \ge 30]$$

Read aloud: *"the sum, for i from 1 to n, of one if x-i is at least 30"*, that is, the number of values that are at least 30. **By hand** for our six essays: 170, 53, 10, 102, 22 and 13 give the indicators 1, 1, 0, 1, 0, 0, which add up to 3.

Python does the same thing: a comparison on an array gives `True` or `False` for each value, and `True` counts as 1. So the sum counts, and the mean gives the share.

````python
print(x >= 30)
print("count:", (x >= 30).sum(), "   share:", (x >= 30).mean())
````

## 8 · Proportions and percentages

A **proportion** is a part divided by its whole, so it lies between 0 and 1. A **percentage** is the same proportion multiplied by 100.

$$p = \frac{\text{part}}{\text{whole}}, \qquad \text{percentage} = 100 \times p$$

Of the journal's 321 articles, 290 are in Italian. The proportion is 290 / 321 = 0.903, or 90.3%. In pandas, `value_counts()` counts each value, and `value_counts(normalize=True)` turns the counts into proportions:

````python
languages = pd.DataFrame({
    "articles": bib.articles["langLabel"].value_counts(),
    "proportion": bib.articles["langLabel"].value_counts(normalize=True).round(3),
})
languages
````

**Percentage points against percent.** When a percentage changes, there are two ways to describe the change, and they are often confused:

- the **difference in percentage points** is the new percentage minus the old one;
- the **relative change in percent** is that difference divided by the old percentage, times 100.

An example from the journal: articles not written in Italian were 4.4% of those published in 2012–2018 and 13.4% of those published in 2019–2025. That is a rise of 9 **percentage points**, and a relative increase of about 200%: the share tripled. Both statements are true; they answer different questions. "Up 9%" would be wrong in both senses.

Move the sliders to see how the two measures relate:

````python
def points_or_percent(old, new):
    fig, ax = plt.subplots(figsize=(5.4, 3))
    ax.bar(["before", "after"], [old, new], color=[plots.MUTED, plots.BLUE])
    ax.set(ylabel="share (%)", ylim=(0, 100), title="Same change, two descriptions")
    plt.show()
    print(f"difference: {new - old:+.1f} percentage points")
    print(f"relative change: {100 * (new - old) / old:+.0f}%")

explore(points_or_percent, old=(1.0, 90.0, 0.5, 4.4, "before (%)"),
        new=(1.0, 90.0, 0.5, 13.4, "after (%)"))
````
